### Overview
This notebook automates the calculation of Principal Stresses ($\sigma_1, \sigma_2, \sigma_3$) and the Von Mises Equivalent Stress ($\sigma_{vm}$) from a $3 \times 3$ stress tensor using `numpy`. It evaluates mechanical yielding against a specified material yield strength.

```python
import numpy as np

def calculate_von_mises_3d(stress_matrix, yield_strength_mpa):
    """
    Computes Principal Stresses, Von Mises Stress, and Factor of Safety (FoS)
    from a 3x3 symmetric stress tensor matrix in MPa.
    """
    stress_tensor = np.array(stress_matrix, dtype=float)
    
    # 1. Verify tensor dimensions
    if stress_tensor.shape != (3, 3):
        raise ValueError("Stress tensor must be a 3x3 matrix.")
    
    # 2. Calculate Principal Stresses (Eigenvalues of the stress matrix)
    eigenvalues = np.linalg.eigvalsh(stress_tensor)
    
    # Sort eigenvalues in descending order: sigma_1 >= sigma_2 >= sigma_3
    sigma_1, sigma_2, sigma_3 = np.sort(eigenvalues)[::-1]
    
    # 3. Compute Von Mises Equivalent Stress
    von_mises = np.sqrt(0.5 * (
        (sigma_1 - sigma_2)**2 + 
        (sigma_2 - sigma_3)**2 + 
        (sigma_3 - sigma_1)**2
    ))
    
    # 4. Factor of Safety
    fos = yield_strength_mpa / von_mises if von_mises > 0 else np.inf
    
    # 5. Output Summary Report
    print("=" * 55)
    print("          STRUCTURAL STRESS ANALYSIS REPORT")
    print("=" * 55)
    print(f"Principal Stress σ1 : {sigma_1:8.2f} MPa")
    print(f"Principal Stress σ2 : {sigma_2:8.2f} MPa")
    print(f"Principal Stress σ3 : {sigma_3:8.2f} MPa")
    print("-" * 55)
    print(f"Von Mises Stress σ_vm: {von_mises:8.2f} MPa")
    print(f"Material Yield Limit: {yield_strength_mpa:8.2f} MPa")
    print(f"Factor of Safety (FoS): {fos:8.2f}")
    print("-" * 55)
    
    if fos >= 1.0:
        print("RESULT: SAFE (No yielding under applied load state)")
    else:
        print("RESULT: FAILURE WARNING (Von Mises stress exceeds yield limit!)")
    print("=" * 55)

    return {
        "sigma_1": sigma_1,
        "sigma_2": sigma_2,
        "sigma_3": sigma_3,
        "von_mises": von_mises,
        "fos": fos
    }

# --- TEST CASE: High-Shear Shaft Element Load State ---
# Stress tensor values in MPa: [sigma_xx, tau_xy, tau_xz], etc.
sample_stress_tensor = [
    [120.0,  45.0,  10.0],
    [ 45.0, -30.0,  15.0],
    [ 10.0,  15.0,   5.0]
]

# Material: Structural Steel (Yield Strength = 250 MPa)
steel_yield_strength = 250.0

# Run calculation
results = calculate_von_mises_3d(sample_stress_tensor, steel_yield_strength)
```
